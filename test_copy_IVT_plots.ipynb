```python
## (a) what is the date you want the model to be valid for?
start_date = '2024-08-25 00'
end_date = '2024-08-26 18'
valid_dates = pd.date_range(start=start_date, end=end_date, freq='6H')
print(valid_dates)
## (b) what is the lead time you want shown?
lead = 0 ## in days
F = lead * 24 # convert to hours
print(F)
## calculate the time of model initialization based on (a) and (b)
yr_lst = []
mon_lst = []
day_lst = []
hr_lst = []
for i, valid_date in enumerate(valid_dates):
    model_init = valid_date - timedelta(hours=float(F))
    yr_lst.append(model_init.strftime("%Y"))
    mon_lst.append(model_init.strftime("%m"))
    day_lst.append(model_init.strftime("%d"))
    hr_lst.append(model_init.strftime("%H"))
    
    # model_init = model_init.strftime("%Y%m%d_%H") # %Y%m%d_%H
    # print(model_init)
    # model_init_lst.append(model_init)


print(yr_lst)
print(mon_lst)
print(day_lst)
print(hr_lst)
```

```python
varname = 'ivt' # 'ivt or '250Wind'
domain = 'NEPac' # 'NEPac'
for i, (yr, mon, day, hr) in enumerate(zip(yr_lst, mon_lst, day_lst, hr_lst)):
    try: 
        fname = "GFS_{5}_{6}_{0}{1}{2}_{3}_F{4}.png".format(yr, mon, day, hr, str(F), varname, domain)
        infile = "/data/downloaded/Forecasts/ARPortal_Archive/gfs/{4}/{5}/{0}{1}{2}/{3}".format(yr, mon, day, fname, varname, domain)
        outfile = "/home/dnash/comet_data/IVT_GFS_maps/{0}".format(fname)
        print(infile, outfile)
        shutil.copyfile(infile, outfile)
    except FileNotFoundError:
        print('Missing {0}'.format(fname))
        pass
```
